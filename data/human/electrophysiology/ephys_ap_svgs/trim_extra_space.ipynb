```python
from svgpathtools import svg2paths2
import matplotlib.pyplot as plt
from matplotlib.path import Path as MplPath
from matplotlib.patches import PathPatch
from xml.etree import ElementTree as ET

def get_y_bounds(paths):
    all_y = []
    for path in paths:
        for seg in path:
            all_y.extend([seg.start.imag, seg.end.imag])
            all_y.extend([seg.end.imag])
    return min(all_y), max(all_y)

def plot_svg_with_bounds(paths, y_min, y_max):
    fig, ax = plt.subplots()

    for path in paths:
        for segment in path:
            x = [segment.start.real, segment.end.real]
            y = [segment.start.imag, segment.end.imag]
            ax.plot(x, y, color='yellowgreen', alpha=0.5)

    # Horizontal bounds
    ax.axhline(y_min, color='red', linestyle='--', label=f'y_min = {y_min:.1f}')
    ax.axhline(y_max, color='blue', linestyle='--', label=f'y_max = {y_max:.1f}')
    ax.legend()
    ax.set_title("SVG with y-bounds")
    ax.set_aspect('equal')
    ax.invert_yaxis()  # Important: SVG Y increases downward
    plt.show()

def crop_svg_y(input_svg, output_svg):
    # Load paths
    paths, attributes, svg_attributes = svg2paths2(input_svg)
    y_min, y_max = get_y_bounds(paths)

    # Show plot
    plot_svg_with_bounds(paths, y_min, y_max)

    # Update viewBox
    tree = ET.parse(input_svg)
    root = tree.getroot()
    ET.register_namespace("", "http://www.w3.org/2000/svg")

    vb = root.attrib.get("viewBox")
    if vb:
        x, y, w, h = map(float, vb.strip().split())
        new_y = y_min
        new_h = y_max - y_min
        root.attrib["viewBox"] = f"{x} {new_y} {w} {new_h}"
        print(f"Updated viewBox: {root.attrib['viewBox']}")
    else:
        print("⚠️ No viewBox found. Skipping crop.")

    tree.write(output_svg)
    print(f"Saved cropped SVG to: {output_svg}")
```

```python
for svg in os.listdir(root):
    if svg.endswith('.svg'):
        crop_svg_y(os.path.join(root, svg), os.path.join(root, svg.replace('.svg', '_trimmed.svg')))
```
