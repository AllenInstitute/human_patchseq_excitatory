```python
import os
import pandas as pd 
from pathlib import Path
```

```python
DATA_ROOT = Path('/root/capsule/data')

reference_umap_file =    DATA_ROOT / 'human' / 'reference' / '20260722' / 'FACS_patchseq_QC_and_UMAP_23June2026.csv'
exclude_from_umap_file = DATA_ROOT / 'human' / 'reference' / '20260722' / 'patchseq_results_23June2026(removed_post_integration).csv'


reference_umap = pd.read_csv(reference_umap_file, index_col=0)
exclude_umap = pd.read_csv(exclude_from_umap_file, index_col=0)
```
