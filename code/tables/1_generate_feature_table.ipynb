# Generate Feature Table Excel Spreadsheet

This notebook creates a multi-tab Excel file with:
- Tab 1: Morphology feature descriptions
- Tab 2: Human morphology dendrite features
- Tab 3: Human morphology axon features
- Tab 4: Macaque morphology dendrite features
- Tab 5: Mouse morphology dendrite features
- Tab 6: Electrophysiology feature descriptions
- Tab 7: Human electrophysiology features
- Tab 8: Macaque electrophysiology features
- Tab 9: Mouse electrophysiology features

```python
import pandas as pd
import numpy as np
```

## Load all data files

```python
# Tab 1: Morphology feature descriptions
tab1_df = pd.read_csv('/root/capsule/data/formatting_dicts/morpho_feature_description.csv')
print(f"Tab 1 (Feature Descriptions): {tab1_df.shape[0]} rows, {tab1_df.shape[1]} columns")

# Create mapping from skeleton_keys_name to feature_name for renaming columns
rename_mapping = dict(zip(tab1_df['skeleton_keys_name'], tab1_df['feature_name']))
print(f"Created rename mapping with {len(rename_mapping)} features")

tab1_df.head()
```

```python
# Tab 2: Human morphology dendrite features
tab2_df = pd.read_csv('/root/capsule/data/human/morphology/dendrite/20260712/RawFeatureWide.csv')

# Rename columns using the mapping (keep specimen_id as is)
tab2_df = tab2_df.rename(columns=rename_mapping)

print(f"\nTab 2 (Human Dendrite): {tab2_df.shape[0]} rows, {tab2_df.shape[1]} columns")
tab2_df.head()
```

```python
# Tab 3: Human morphology axon features
tab3_df = pd.read_csv('/root/capsule/data/human/morphology/axon/20260715/RawFeatureWide.csv')

# Rename columns using the mapping (keep specimen_id as is)
tab3_df = tab3_df.rename(columns=rename_mapping)

print(f"\nTab 3 (Human Axon): {tab3_df.shape[0]} rows, {tab3_df.shape[1]} columns")
tab3_df.head()
```

```python
# Tab 4: Macaque morphology dendrite features
tab4_df = pd.read_csv('/root/capsule/data/macaque/morphology/dendrite/20260722/RawFeatureWide.csv')

# Rename columns using the mapping (keep specimen_id as is)
tab4_df = tab4_df.rename(columns=rename_mapping)

print(f"\nTab 4 (Macaque Dendrite): {tab4_df.shape[0]} rows, {tab4_df.shape[1]} columns")
tab4_df.head()
```

```python
# Tab 5: Mouse morphology dendrite features (from combined file)
# Load mouse metadata to filter for restricted analysis cells
mouse_metadata = pd.read_csv('/root/capsule/data/mouse/metadata/20260826/mouse_metadata.csv')
restricted_cells = mouse_metadata[mouse_metadata['restricted analysis?'] == True]['specimen_id'].astype(str).tolist()
print(f"Restricting mouse data to {len(restricted_cells)} cells in restricted analysis")

mouse_combined_df = pd.read_csv('/root/capsule/data/mouse/morphology/dendrite/20260826/RawFeatureWide.csv')

# Filter to only restricted analysis cells
mouse_combined_df = mouse_combined_df[mouse_combined_df['specimen_id'].astype(str).isin(restricted_cells)]
print(f"Filtered mouse data from {len(pd.read_csv('/root/capsule/data/mouse/morphology/dendrite/20260826/RawFeatureWide.csv'))} to {len(mouse_combined_df)} cells")

# Separate dendrite and axon columns
# Dendrite columns: include apical, basal dendrite features + soma + basal stem exit + axon exit
dendrite_cols = ['specimen_id'] + [col for col in mouse_combined_df.columns 
                                    if col != 'specimen_id' and 'axon' not in col.lower()]
# Also include axon_exit features in dendrite tab (to match human structure)
dendrite_cols += [col for col in mouse_combined_df.columns if col.startswith('axon_exit')]

# Create dendrite dataframe
tab5_df = mouse_combined_df[dendrite_cols].copy()

# Rename columns using the mapping
tab5_df = tab5_df.rename(columns=rename_mapping)

print(f"\nTab 5 (Mouse Dendrite): {tab5_df.shape[0]} rows, {tab5_df.shape[1]} columns")
tab5_df.head()
```

```python
# Tab 6: Electrophysiology feature descriptions
ephys_rename_df = pd.read_csv('/root/capsule/data/formatting_dicts/ephys_feature_rename.csv')

# Create description dataframe with feature_name, ipfx_name, and empty description
tab7_df = pd.DataFrame({
    'feature_name': ephys_rename_df['new_feature_name'],
    'ipfx_name': ephys_rename_df['old_feature_name'],
    'description': ''
})

# Drop duplicate feature_name entries (keep first)
tab7_df = tab7_df.drop_duplicates(subset=['feature_name'], keep='first')

# Create mapping from old_feature_name (ipfx_name) to new_feature_name for renaming columns
ephys_rename_mapping = dict(zip(ephys_rename_df['old_feature_name'], ephys_rename_df['new_feature_name']))

# Get list of valid feature names for filtering later tabs
valid_ephys_features = tab7_df['feature_name'].tolist()

print(f"\nTab 6 (Ephys Feature Descriptions): {tab7_df.shape[0]} rows, {tab7_df.shape[1]} columns")
print(f"Created ephys rename mapping with {len(ephys_rename_mapping)} features")
tab7_df.head()
```

```python
# Tab 7: Human electrophysiology features
tab8_df = pd.read_csv('/root/capsule/data/human/electrophysiology/features/20260807/ephys_features.csv')

# Rename columns using the mapping
tab8_df = tab8_df.rename(columns=ephys_rename_mapping)

# Filter to only include specimen_id and columns that are in Tab 6
tab8_cols = ['specimen_id'] + [col for col in tab8_df.columns if col in valid_ephys_features]
tab8_df = tab8_df[tab8_cols]

print(f"\nTab 7 (Human Ephys): {tab8_df.shape[0]} rows, {tab8_df.shape[1]} columns")
tab8_df.head()
```

```python
# Tab 8: Macaque electrophysiology features
tab9_df = pd.read_csv('/root/capsule/data/macaque/electrophysiology/20260908/ephys_features.csv')

# Rename columns using the mapping
tab9_df = tab9_df.rename(columns=ephys_rename_mapping)

# Filter to only include specimen_id and columns that are in Tab 6
tab9_cols = ['specimen_id'] + [col for col in tab9_df.columns if col in valid_ephys_features]
tab9_df = tab9_df[tab9_cols]

print(f"\nTab 8 (Macaque Ephys): {tab9_df.shape[0]} rows, {tab9_df.shape[1]} columns")
tab9_df.head()
```

```python
# Tab 9: Mouse electrophysiology features
tab10_df = pd.read_csv('/root/capsule/data/mouse/electrophysiology/20260908/ephys_features.csv')

# Filter to only restricted analysis cells (same 23 cells as morphology)
tab10_df = tab10_df[tab10_df['specimen_id'].astype(str).isin(restricted_cells)]
print(f"Filtered mouse ephys data to {len(tab10_df)} cells (restricted analysis)")

# Rename columns using the mapping
tab10_df = tab10_df.rename(columns=ephys_rename_mapping)

# Filter to only include specimen_id and columns that are in Tab 6
tab10_cols = ['specimen_id'] + [col for col in tab10_df.columns if col in valid_ephys_features]
tab10_df = tab10_df[tab10_cols]

# Remove columns that are entirely NaN (no data for any of the restricted cells)
cols_before = len(tab10_df.columns)
feature_cols = [col for col in tab10_df.columns if col != 'specimen_id' and tab10_df[col].notna().any()]
tab10_df = tab10_df[['specimen_id'] + feature_cols]
cols_removed = cols_before - len(tab10_df.columns)
if cols_removed > 0:
    print(f"Removed {cols_removed} columns with all NaN values")

print(f"\nTab 9 (Mouse Ephys): {tab10_df.shape[0]} rows, {tab10_df.shape[1]} columns")
tab10_df.head()
```

## Create Excel file with all tabs

```python
# Write all tabs to Excel file
output_path = '/root/capsule/scratch/20260807/tables/feature_table.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    tab1_df.to_excel(writer, sheet_name='morphology_feature_description', index=False)
    tab2_df.to_excel(writer, sheet_name='human_dendrite', index=False)
    tab3_df.to_excel(writer, sheet_name='human_axon', index=False)
    tab4_df.to_excel(writer, sheet_name='macaque_dendrite', index=False)
    tab5_df.to_excel(writer, sheet_name='mouse_dendrite', index=False)
    tab7_df.to_excel(writer, sheet_name='ephys_feature_description', index=False)
    tab8_df.to_excel(writer, sheet_name='human_ephys', index=False)
    tab9_df.to_excel(writer, sheet_name='macaque_ephys', index=False)
    tab10_df.to_excel(writer, sheet_name='mouse_ephys', index=False)

print(f"\nSuccessfully created Excel file with 9 tabs at: {output_path}")
print("\nTab summary:")
print(f"  Tab 1 - morphology_feature_description: {tab1_df.shape[0]} rows, {tab1_df.shape[1]} columns")
print(f"  Tab 2 - human_dendrite: {tab2_df.shape[0]} rows, {tab2_df.shape[1]} columns")
print(f"  Tab 3 - human_axon: {tab3_df.shape[0]} rows, {tab3_df.shape[1]} columns")
print(f"  Tab 4 - macaque_dendrite: {tab4_df.shape[0]} rows, {tab4_df.shape[1]} columns")
print(f"  Tab 5 - mouse_dendrite: {tab5_df.shape[0]} rows, {tab5_df.shape[1]} columns")
print(f"  Tab 6 - ephys_feature_description: {tab7_df.shape[0]} rows, {tab7_df.shape[1]} columns")
print(f"  Tab 7 - human_ephys: {tab8_df.shape[0]} rows, {tab8_df.shape[1]} columns")
print(f"  Tab 8 - macaque_ephys: {tab9_df.shape[0]} rows, {tab9_df.shape[1]} columns")
print(f"  Tab 9 - mouse_ephys: {tab10_df.shape[0]} rows, {tab10_df.shape[1]} columns")
```
