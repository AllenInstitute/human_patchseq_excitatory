```python
# Configuration & Helper Functions

# Constants
MICRON_SCALING_FACTOR = 1.144
MS_CONVERSION_FACTOR = 1000

def apply_taxonomy_corrections(df, subclass_col, cluster_col, L23_rename_dict, ttype_layer_dict):
    """
    Apply standard taxonomy corrections to metadata.

    Operations:
    1. L5 ET standardization ('L5 ET_ME' → 'L5 ET', 'L5 ET(ME)' → 'L5 ET')
    2. Apply L23 type renaming
    3. Map cluster to layer

    Parameters
    ----------
    df : pandas.DataFrame
        Metadata dataframe to transform
    subclass_col : str
        Name of subclass column
    cluster_col : str
        Name of cluster/type column
    L23_rename_dict : dict
        Dictionary mapping L2/3 type names
    ttype_layer_dict : dict
        Dictionary mapping types to layers

    Returns
    -------
    pandas.DataFrame
        Modified dataframe (in-place operations)
    """
    # L5 ET standardization
    df[subclass_col] = df[subclass_col].replace('L5 ET_ME', 'L5 ET').replace('L5 ET(ME)', 'L5 ET')
    df[cluster_col] = df[cluster_col].replace('L5 ET_ME', 'L5 ET(ME)')

    # Apply L23 renaming
    df[cluster_col] = df[cluster_col].apply(lambda x: L23_rename_dict.get(x, x))

    # Map to layer
    df['layer'] = df[cluster_col].map(ttype_layer_dict)

    return df


def split_layer_2_3(df, layer_col='layer', subclass_col='subclass_label'):
    """
    Add layer_subclass and layer_split columns for Layer 2/3 splitting.

    Rules:
    - Layer 2 → layer_subclass='L2 IT', layer_split='L2'
    - Layer 3 → layer_subclass='L3 IT', layer_split='L3'
    - Other layers → use subclass_col and layer as-is

    Parameters
    ----------
    df : pandas.DataFrame
        Dataframe with layer information
    layer_col : str
        Name of layer column
    subclass_col : str
        Name of subclass column

    Returns
    -------
    pandas.DataFrame
        Modified dataframe with new columns
    """
    df["layer_subclass"] = df.apply(
        lambda row: "L2 IT" if row[layer_col] == 'Layer 2' else
                   ("L3 IT" if row[layer_col] == 'Layer 3' else row[subclass_col]),
        axis=1
    )
    df["layer_split"] = df.apply(
        lambda row: "L2" if row[layer_col] == 'Layer 2' else
                   ("L3" if row[layer_col] == 'Layer 3' else row[layer_col]),
        axis=1
    )
    return df


def add_layer_grouping(df, subclass_col='subclass_label'):
    """
    Derive layer and layer_group from subclass.

    - layer: 'L2/3', 'L4', 'L5', 'L6' extracted from subclass prefix
    - layer_group: 'L2/3/4' or 'L5/6' grouping

    Parameters
    ----------
    df : pandas.DataFrame
        Dataframe with subclass information
    subclass_col : str
        Name of subclass column

    Returns
    -------
    pandas.DataFrame
        Modified dataframe with layer and layer_group columns
    """
    # Extract layer from subclass
    df['layer'] = df[subclass_col].apply(
        lambda x: next((layer for layer in ['L2/3', 'L4', 'L5', 'L6'] if str(x).startswith(layer)), None)
    )

    # Map to layer group
    layer_group_mapping = {'L2/3': 'L2/3/4', 'L4': 'L2/3/4', 'L5': 'L5/6', 'L6': 'L5/6'}
    df['layer_group'] = df['layer'].map(layer_group_mapping)

    return df


def apply_lobe_filter_columns(df, t_type_col='t_type', subclass_col='subclass_label',
                               lobe_col='lobe', keep_lobe='TemL'):
    """
    Create filtered _teml variants: set to 'gray' if lobe != keep_lobe.

    Adds: t_type_teml, subclass_label_teml columns

    Parameters
    ----------
    df : pandas.DataFrame
        Dataframe with lobe information
    t_type_col : str
        Name of type column
    subclass_col : str
        Name of subclass column
    lobe_col : str
        Name of lobe column
    keep_lobe : str
        Lobe to keep (others set to 'gray')

    Returns
    -------
    pandas.DataFrame
        Modified dataframe with filtered columns
    """
    def color_label(row, col):
        if row[lobe_col] != keep_lobe:
            return 'gray'
        else:
            return row[col]

    df[f'{t_type_col}_teml'] = df.apply(color_label, axis=1, args=(t_type_col,))
    df[f'{subclass_col}_teml'] = df.apply(color_label, axis=1, args=(subclass_col,))

    return df


flatten = lambda l: list(chain.from_iterable(l))
```

```python
features = raw_feature_wide_dend.columns.tolist()[1:]

# subclass x feature stats
group_var = 'layer'
group_var_order = list(subclass_order_dict.keys())

morpho_eta_squared_group = dict()
morpho_kw_pvals_fdr_group = dict()
morpho_dunn_pvals_fdr_group = dict()

for group in group_var_order:
    print(group)
    group_data = morpho_sorted_clean_data_dend.loc[morpho_sorted_clean_data_dend[group_var] == group]

    # Get t-types in the current group
    x_var = 't_type'
    x_var_order = flatten([tx_order_dict[subclass] for subclass in group_data.subclass_label.unique()])

    # Initialize dictionaries for Kruskal-Wallis H test and p-values
    kw_hvals = dict()
    kw_pvals = dict()

    # Iterate over features to calculate KW test and eta-squared
    for f in features:
        # Get groups of data for each t-type
        t_groups = [group_data.loc[group_data[x_var] == x][f].dropna() for x in x_var_order]

        # Only include non-empty groups with at least 1 cell
        t_groups = [g for g in t_groups if len(g) >= 1]

        if len(t_groups) < 2:
            print(f"Not enough t_groups with data for feature {f}")
            continue

        try:
            hvalue, pvalue = stats.kruskal(*t_groups, nan_policy='omit')
            kw_hvals[f] = hvalue
            kw_pvals[f] = pvalue
        except ValueError as e:
            print(f'Kruskal-Wallis not calculated for {f} because of an error: {e}')

    # FDR correction for KW p-values
    kw_pvals_list = list(kw_pvals.values())
    kw_rejected_list, kw_pvals_fdr_list = fdrcorrection(kw_pvals_list, alpha=0.05, method='indep', is_sorted=False)

    kw_pvals_fdr = dict(zip(list(kw_pvals.keys()), kw_pvals_fdr_list))
    kw_rejected = dict(zip(list(kw_pvals.keys()), kw_rejected_list))

    # Post-hoc Dunn test for significant KW features (after FDR correction)
    dunn_pvals_fdr = dict()
    for k, v in kw_rejected.items():
        if v:  # only if null hypothesis was rejected
            dunn_pvals_fdr[k] = posthoc_dunn(group_data[group_data[x_var].isin(x_var_order)], val_col=k, group_col=x_var, p_adjust='fdr_bh', sort=True)

    # Calculate eta-squared for the KW H-test
    eta_squared_ttypes = {}
    k = len(t_groups)  # number of t-types (groups)
    n = sum([len(g) for g in t_groups])  # total number of observations
    print(f'k: {k}, n: {n}')
    for f, h in kw_hvals.items():
        esq = (h - k + 1) / (n - k)  # eta^2 formula for KW test
        eta_squared_ttypes[f] = esq

    eta_squared_ttypes_df = pd.DataFrame.from_dict(sorted(eta_squared_ttypes.items(), key=operator.itemgetter(1), reverse=True))
    eta_squared_ttypes_df.rename(columns={0: 'feature', 1: 'eta_squared'}, inplace=True)

    # Store results
    morpho_eta_squared_group[group] = eta_squared_ttypes_df
    morpho_kw_pvals_fdr_group[group] = kw_pvals_fdr
    morpho_dunn_pvals_fdr_group[group] = dunn_pvals_fdr
```

```python
#prep for viewer section
# num_cells_per_ttype = 5

morpho_viewer_cells = {}
for layer in layers:
    sorted_cells_ttype = morpho_sorted_clean_data_dend[morpho_sorted_clean_data_dend['layer_split'] == layer]

    def sample_representative(group, n=layer_num_cells_per_ttype[layer]):
        teml_cells = group[group['lobe'] == 'TemL'].sort_values('soma_distance_from_pia')

        # Helper function: evenly sample across soma depth
        def even_sample(df, n):
            if len(df) <= n:
                return df
            return df.iloc[np.linspace(0, len(df) - 1, n, dtype=int)]

        if len(teml_cells) >= n:
            sampled = even_sample(teml_cells, n)
        else:
            # Take all TemL cells
            sampled = teml_cells.copy()
            remaining = n - len(sampled)

            other_cells = group[group['lobe'] != 'TemL'].sort_values('soma_distance_from_pia')
            sampled_others = even_sample(other_cells, remaining)
            sampled = pd.concat([sampled, sampled_others], ignore_index=True)

        return sampled

    temp = (
        sorted_cells_ttype
        .reset_index()
        .groupby('t_type', group_keys=False)
        .apply(sample_representative)
        .reset_index(drop=True)
    )

    temp['t_type'] = pd.Categorical(temp['t_type'], categories=TTYPE_ORDER, ordered=True)

    # sort by t_type and soma depth for final order
    morpho_viewer_cells[layer] = temp.sort_values(by=['t_type', 'soma_distance_from_pia'])
```

Morpho Eta2 and FDR corrected KW of ttypes within each subclass

```python
features = raw_feature_wide_dend.columns.tolist()[1:]

# subclass x feature stats
group_var = 'layer_split'
group_var_order = layers

morpho_eta_squared_group = dict()
morpho_kw_pvals_fdr_group = dict()
morpho_dunn_pvals_fdr_group = dict()

for group in group_var_order:
    print(group)
    group_data = morpho_sorted_clean_data_dend.loc[morpho_sorted_clean_data_dend[group_var] == group]

    # Get t-types in the current group
    x_var = 't_type'
    x_var_order = [t for t in TTYPE_ORDER if t in group_data[x_var].unique()]

    # Initialize dictionaries for Kruskal-Wallis H test and p-values
    kw_hvals = dict()
    kw_pvals = dict()

    # Iterate over features to calculate KW test and eta-squared
    for f in features:
        # Get groups of data for each t-type
        t_groups = [group_data.loc[group_data[x_var] == x][f].dropna() for x in x_var_order]

        # Only include non-empty groups with at least 1 cell
        t_groups = [g for g in t_groups if len(g) >= 1]

        if len(t_groups) < 2:
            print(f"Not enough t_groups with data for feature {f}")
            continue

        try:
            hvalue, pvalue = stats.kruskal(*t_groups, nan_policy='omit')
            kw_hvals[f] = hvalue
            kw_pvals[f] = pvalue
        except ValueError as e:
            print(f'Kruskal-Wallis not calculated for {f} because of an error: {e}')

    # FDR correction for KW p-values
    kw_pvals_list = list(kw_pvals.values())
    kw_rejected_list, kw_pvals_fdr_list = fdrcorrection(kw_pvals_list, alpha=0.05, method='indep', is_sorted=False)

    kw_pvals_fdr = dict(zip(list(kw_pvals.keys()), kw_pvals_fdr_list))
    kw_rejected = dict(zip(list(kw_pvals.keys()), kw_rejected_list))

    # Post-hoc Dunn test for significant KW features (after FDR correction)
    dunn_pvals_fdr = dict()
    for k, v in kw_rejected.items():
        if v:  # only if null hypothesis was rejected
            dunn_pvals_fdr[k] = posthoc_dunn(group_data[group_data[x_var].isin(x_var_order)], val_col=k, group_col=x_var, p_adjust='fdr_bh', sort=True)

    # Calculate eta-squared for the KW H-test
    eta_squared_ttypes = {}
    k = len(t_groups)  # number of t-types (groups)
    n = sum([len(g) for g in t_groups])  # total number of observations
    print(f'k: {k}, n: {n}')
    for f, h in kw_hvals.items():
        esq = (h - k + 1) / (n - k)  # eta^2 formula for KW test
        eta_squared_ttypes[f] = esq

    eta_squared_ttypes_df = pd.DataFrame.from_dict(sorted(eta_squared_ttypes.items(), key=operator.itemgetter(1), reverse=True))
    eta_squared_ttypes_df.rename(columns={0: 'feature', 1: 'eta_squared'}, inplace=True)

    # Store results
    morpho_eta_squared_group[group] = eta_squared_ttypes_df
    morpho_kw_pvals_fdr_group[group] = kw_pvals_fdr
    morpho_dunn_pvals_fdr_group[group] = dunn_pvals_fdr
```
