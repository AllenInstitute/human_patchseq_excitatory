# Consolidate Statistics

This notebook consolidates various statistical analyses into a single Excel workbook with multiple tabs.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Output file path
scratch_dir = Path('/root/capsule/scratch/20260807')
output_dir = scratch_dir / 'statistics_table'
output_file = output_dir / 'S3_statistics_table.xlsx'
```

```python
# Initialize Excel writer
writer = pd.ExcelWriter(output_file, engine='openpyxl')
```

## Tab 1: Morphology Subclass Effect Sizes

```python
# Load morphology subclass effect sizes
df_morpho = pd.read_csv('/root/capsule/scratch/20260807/effect_sizes/morphology_subclass_effect_sizes.csv')
print(f"Morphology shape: {df_morpho.shape}")
df_morpho.head()
```

```python
# Write to Excel - Tab 1
df_morpho.to_excel(writer, sheet_name='M_effect_size', index=False)
print("✓ Tab 1: M_effect_size added")
```

## Tab 2: Electrophysiology Subclass Effect Sizes

```python
# Load electrophysiology subclass effect sizes
df_ephys = pd.read_csv('/root/capsule/scratch/20260807/effect_sizes/electrophysiology_subclass_effect_sizes.csv')
print(f"Electrophysiology shape: {df_ephys.shape}")
df_ephys.head()
```

```python
# Write to Excel - Tab 2
df_ephys.to_excel(writer, sheet_name='E_effect_size', index=False)
print("✓ Tab 2: E_effect_size added")
```

## Tab 3: Dendrite Subclass Statistics (within layer)

```python
# Load dendrite statistics
df_dendrite = pd.read_csv('/root/capsule/scratch/20260807/dendrite/human_excitatory_boxplots_dendrite_features_KW_Dunn_FDR_bySubclass.csv')
print(f"Dendrite stats shape: {df_dendrite.shape}")
df_dendrite.head()
```

```python
# Write to Excel - Tab 3
df_dendrite.to_excel(writer, sheet_name='M_dend_subclass_stats', index=False)
print("✓ Tab 3: M_dend_subclass_stats added")
```

## Tab 4: Axon Subclass Statistics (within layer)

```python
# Load axon statistics
df_axon = pd.read_csv('/root/capsule/scratch/20260807/axon/human_excitatory_boxplots_axon_features_KW_Dunn_FDR_bySubclass.csv')
print(f"Axon stats shape: {df_axon.shape}")
df_axon.head()
```

```python
# Write to Excel - Tab 4
df_axon.to_excel(writer, sheet_name='M_axon_subclass_stats', index=False)
print("✓ Tab 4: M_axon_subclass_stats added")
```

## Tab 5: Electrophysiology Subclass Statistics (within layer)

```python
# Load ephys statistics
df_ephys_stats = pd.read_csv('/root/capsule/scratch/20260807/ephys/human_excitatory_boxplots_ephys_features_KW_Dunn_FDR_bySubclass.csv')
print(f"Ephys stats shape: {df_ephys_stats.shape}")
df_ephys_stats.head()
```
