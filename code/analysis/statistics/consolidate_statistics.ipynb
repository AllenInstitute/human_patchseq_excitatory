## Tab 14: Dprime subclass

```python
def process_subclass_dprime(df, modality):
    df['modality'] = modality
    return df[['modality', 'subclass_label', 'dprime', 'cluster_1', 'cluster_2']]
```

```python
df_m = pd.read_csv('/root/capsule/scratch/20260807/dprime/dprime_subclass_M.csv')
df_e = pd.read_csv('/root/capsule/scratch/20260807/dprime/dprime_subclass_E.csv')
df_me = pd.read_csv('/root/capsule/scratch/20260807/dprime/dprime_subclass_ME.csv')

df_m = process_subclass_dprime(df_m, 'M')
df_e = process_subclass_dprime(df_e, 'E')
df_me = process_subclass_dprime(df_me, 'ME')

df = pd.concat([df_m, df_e, df_me])
df
```

```python
# Write to Excel 
df.to_excel(writer, sheet_name='dprime_subclass', index=False)
print("✓ Tab: dprime_subclass added")
```
