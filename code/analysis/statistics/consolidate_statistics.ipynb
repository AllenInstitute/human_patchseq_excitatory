# Consolidate Statistics

This notebook consolidates various statistical analyses into a single Excel workbook with multiple tabs.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Output file path
scratch_dir = Path('/root/capsule/scratch/20260807')
output_dir = scratch_dir / 'statistics_table'
output_file = output_dir / 'S3_statistics_table.xlsx'
```

```python
# Initialize Excel writer
writer = pd.ExcelWriter(output_file, engine='openpyxl')
```

## Tab 1: Morphology Subclass Effect Sizes

```python
# Load morphology subclass effect sizes
df_morpho = pd.read_csv('/root/capsule/scratch/20260807/effect_sizes/morphology_subclass_effect_sizes.csv')
print(f"Morphology shape: {df_morpho.shape}")
df_morpho.head()
```

```python
# Write to Excel - Tab 1
df_morpho.to_excel(writer, sheet_name='M_effect_size', index=False)
```

## Tab 2: Electrophysiology Subclass Effect Sizes

```python
# Load electrophysiology subclass effect sizes
df_ephys = pd.read_csv('/root/capsule/scratch/20260807/effect_sizes/electrophysiology_subclass_effect_sizes.csv')
print(f"Electrophysiology shape: {df_ephys.shape}")
df_ephys.head()
```

```python
# Write to Excel - Tab 2
df_ephys.to_excel(writer, sheet_name='E_effect_size', index=False)
```

```python
# Save and close the Excel file
writer.close()
print(f"✓ Excel file saved to: {output_file}")
```
