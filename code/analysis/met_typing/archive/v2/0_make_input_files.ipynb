Make my data inputs

```python
out_dir = r'\\allen\programs\celltypes\workgroups\mousecelltypes\SarahWB\datasets\human_exc\analysis\20250815_seaad\MET_typing\data'

e_data = pd.read_csv(r"\\allen\programs\celltypes\workgroups\mousecelltypes\SarahWB\datasets\human_exc\data\ephys\20250812\sparse_pca_components_exc_human_May2025_flex.csv", index_col=0)
m_data = pd.read_csv(r"\\allen\programs\celltypes\workgroups\mousecelltypes\SarahWB\datasets\human_exc\data\Features\dendrite\NormFeatureWide.csv") #use the normalizezd morph data - that's what NG does in exc_vis_manuscript on Github: https://github.com/AllenInstitute/exc_vis_manuscript/blob/main/scripts/input_json/exc_met_mouse_me_clustering.json
metadata = pd.read_csv(r'\\allen\programs\celltypes\workgroups\mousecelltypes\SarahWB\datasets\human_exc\data\metadata\20250804\Datalock_Human_Glut_4_4_25(patchseq_mapping_results).csv', encoding='latin1')
metadata_met = metadata[(metadata['Transcriptomics '] == True) & (metadata['ephys?'] == True) & (metadata['Morphology'] == 'TRUE')]

met_ids = list(set(metadata_met.cell_id).intersection(set(e_data.index)).intersection(set(m_data.specimen_id)))
len(met_ids)
```

Ephys

```python
e_data = e_data[e_data.index.isin(met_ids)]
e_data
```

```python
e_data.to_csv(os.path.join(out_dir, 'e_data.csv'))
```

Morphology

```python
m_data = m_data[m_data.specimen_id.isin(met_ids)]
m_data
```

```python
m_data.to_csv(os.path.join(out_dir, 'm_data.csv'), index=False)
```

Transcriptomics

```python
t_data = metadata[metadata.cell_id.isin(met_ids)]
t_data
```

```python
#add color col
color_dict = get_seaad_colors()
t_data['cluster_color'] = t_data['cluster_label_Hierarchical'].map(lambda x: color_dict[x])

#format cols names & types 
t_data = t_data.rename(columns={'cell_id':'spec_id_label',
                                    'exp_component_name':'sample_id', 
                                    'cluster_label_Hierarchical':'cluster_label'})
t_data['spec_id_label'] = t_data['spec_id_label'].astype(int)

t_data[['spec_id_label', 'sample_id', 'cluster_label', 'cluster_color']]
```

```python
t_data.reset_index(drop=True).to_feather(os.path.join(out_dir, 't_data.feather'))
t_data.to_csv(os.path.join(out_dir, 'metadata.csv'), index=False)
```
