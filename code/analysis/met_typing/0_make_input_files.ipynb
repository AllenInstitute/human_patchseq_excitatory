```python
import os 
import json
import pandas as pd
from pathlib import Path

import sys 
sys.path.append('../../figures') 
from core_utils import get_seaad_colors
```

Make my data inputs

```python
# Save data root 
out_dir = Path('/root/capsule/scratch') / '20260807' / 'MET_typing' / 'data'
os.makedirs(out_dir, exist_ok=True)

# Data paths
DATA_ROOT = Path('/root/capsule/data')

metadata_path = DATA_ROOT / 'human' / 'metadata'   / '20260807' / 'human_metadata.csv'
e_file =        DATA_ROOT / 'human' / 'electrophysiology' / 'sparce_pca' / '20260807' / 'ephys_sparse_pca.csv'
m_file =        DATA_ROOT / 'human' / 'morphology' / 'dendrite' / '20260712' / 'NormFeatureWide.csv'

#load data
e_data = pd.read_csv(e_file, index_col=0)
m_data = pd.read_csv(m_file) 
metadata = pd.read_csv(metadata_path)

#get triple modality cells
metadata_met = metadata[(metadata['Transcriptomics?'] == True) & (metadata['Electrophysiology'] == True) & (metadata['Morphology'] == True)]

met_ids = list(set(metadata_met.cell_id).intersection(set(e_data.index)).intersection(set(m_data.specimen_id)))
len(met_ids)
```

Ephys

```python
e_data = e_data[e_data.index.isin(met_ids)]
e_data
```

```python
e_data.to_csv(os.path.join(out_dir, 'e_data.csv'))
```

Morphology

```python
m_data = m_data[m_data.specimen_id.isin(met_ids)]
m_data
```

```python
m_data.to_csv(os.path.join(out_dir, 'm_data.csv'), index=False)
```

Transcriptomics

```python
t_data = metadata[metadata.cell_id.isin(met_ids)]
t_data
```

```python
#add color col
color_dict = get_seaad_colors()
t_data['cluster_color'] = t_data['cluster_label_Hierarchical'].map(lambda x: color_dict[x])

#format cols names & types 
t_data = t_data.rename(columns={'cell_id':'spec_id_label',
                                    'exp_component_name':'sample_id', 
                                    'cluster_label_Hierarchical':'cluster_label'})
t_data['spec_id_label'] = t_data['spec_id_label'].astype(int)

t_data[['spec_id_label', 'sample_id', 'cluster_label', 'cluster_color']]
```

```python
t_data.reset_index(drop=True).to_feather(os.path.join(out_dir, 't_data.feather'))
t_data.to_csv(os.path.join(out_dir, 'metadata.csv'), index=False)
```
